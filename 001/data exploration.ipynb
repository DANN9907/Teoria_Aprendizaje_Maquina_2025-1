```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Cargar dataset (ajusta la ruta según tu entorno)
df = pd.read_csv("1F96863_3.4.1 - cardiovascular disease, cancer, diabetes or chronic respiratory disease\\170_1F96863_Dataset_2024-12-18.csv")

# Filtrar solo las columnas útiles
df = df[[
    'DIM_GEO_CODE_M49',
    'GEO_NAME_SHORT',
    'DIM_TIME',
    'Sex',
    'RATE_PER_100_N',
    'RATE_PER_100_NL',
    'RATE_PER_100_NU'
]]

# Configuración general de estilo
sns.set(style="whitegrid")

# === Gráfico de tendencia temporal por sexo ===
plt.figure(figsize=(10, 6))
sns.lineplot(data=df, x="DIM_TIME", y="RATE_PER_100_N", hue="Sex", marker="o")
plt.title("Tasa de mortalidad prematura por sexo (por cada 100 personas)")
plt.xlabel("Año")
plt.ylabel("Tasa de mortalidad prematura")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# === Boxplot por sexo ===
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Sex", y="RATE_PER_100_N")
plt.title("Distribución de la tasa de mortalidad prematura por sexo")
plt.xlabel("Sexo")
plt.ylabel("Tasa de mortalidad prematura")
plt.tight_layout()
plt.show()

# === Estadísticas descriptivas por sexo ===
df.info()
```

```python
# Cargar el dataset (ajusta la ruta si es necesario)
df = pd.read_csv("E0D4E17_5.2.2 - Sexual violence by persons other than an intimate partner (last 12 months)\\170_E0D4E17_Dataset_2024-01-08.csv")

# Extraer valores
año = df["DIM_TIME"].iloc[0]
valor = df["RATE_PER_100_N"].iloc[0]
inferior = df["RATE_PER_100_NL"].iloc[0]
superior = df["RATE_PER_100_NU"].iloc[0]

# Crear gráfico de barra con intervalo
plt.figure(figsize=(6, 4))
plt.bar([año], [valor], yerr=[[valor - inferior], [superior - valor]], capsize=10, color='crimson')
plt.title("Prevalencia de violencia por parte de la pareja (Colombia, 2018)")
plt.ylabel("Porcentaje (%)")
plt.ylim(0, 50)
plt.xticks([año], [str(año)])
plt.tight_layout()
plt.show()

# Mostrar el dato directamente
print(f"Año: {año} - Prevalencia estimada: {valor}%, IC: [{inferior}%, {superior}%]")

df.info()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Cargar el dataset (ajusta la ruta si es necesario)
df = pd.read_csv("ED50112_3.9.2 - Unsafe water, unsafe sanitation and lack of hygiene\\170_ED50112_Dataset_2024-01-08.csv")

# Extraer los valores
año = df["DIM_TIME"].iloc[0]
valor = df["RATE_PER_100000_N"].iloc[0]
sexo = df["Sex"].iloc[0]

# Crear gráfico de barra
plt.figure(figsize=(6, 4))
plt.bar([f"{sexo} - {año}"], [valor], color="steelblue")
plt.title("Muertes atribuibles a agua, saneamiento e higiene inseguros (Colombia)")
plt.ylabel("Tasa por cada 100,000 habitantes")
plt.ylim(0, max(5, valor + 1))
plt.tight_layout()
plt.show()

# Imprimir el valor
print(f"Año: {año}, Sexo: {sexo}, Tasa de muertes por WASH: {valor} por cada 100,000 habitantes")

df.info()
```
